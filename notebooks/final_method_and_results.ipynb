## Plain-Language Overview

The challenge was to find strong input settings for eight hidden functions. Each function returned only one score for each submitted input, and larger scores were better. Because the true equations were hidden and the query budget was small, the approach used careful iteration rather than blind search. Early rounds explored promising areas, while later rounds focused on local improvement around the best observed regions. The final workflow combined candidate-generation scripts, sanity checks, classifier checks, historical backtests, and manual review before each submission.

## Method Summary

The final approach was a state-policy trust-region workflow. Each function was treated according to its recent state: continue local exploitation when momentum was strong, refine around the historical best after small misses, reset to the best basin after poor exploration, and allow only bounded alternatives when a function was stagnant. Candidate generation used Gaussian-process-style search for lower-dimensional functions and local random-forest guidance for higher-dimensional functions. Logistic regression, RBF SVM, MLP surrogate checks, distance checks, and historical backtests were used as secondary evidence rather than automatic decision-makers.
